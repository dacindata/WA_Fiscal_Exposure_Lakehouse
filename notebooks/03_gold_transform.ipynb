```python
%sql
Create or replace table wa_fiscal.gold.price_vs_assumption as
SELECT 
    m.month_start,
    m.price_usd_per_tonne,
    m.price_aud_per_tonne,
    m.aud_usd,    
    m.aud_inr,
    a.assumed_price_usd,
    m.price_usd_per_tonne-a.assumed_price_usd as variance_usd,
    (m.price_usd_per_tonne-a.assumed_price_usd) * a.sensitivity_aud_m_per_usd as implied_revenue_impact_aud_m
FROM wa_fiscal.silver.iron_ore_monthly m 
cross join wa_fiscal.bronze.treasury_assumptions a
where a.commodity = 'iron_ore';
```
