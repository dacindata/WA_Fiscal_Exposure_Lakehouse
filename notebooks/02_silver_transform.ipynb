```python
%sql
Create or replace table wa_fiscal.silver.iron_ore_monthly as
with p AS(
    SELECT 
        to_date(concat(substr(period,1,4),'-',substr(period,6,2),'-01')) as month_start,
        CAST(price_usd_per_tonne as double) as price_usd_per_tonne
    from wa_fiscal.bronze.commodity_prices
    where price_usd_per_tonne is not null
),
f as (
    SELECT 
        date_trunc('month',to_date(period,'yyyy-MM-dd')) as month_start,
        CAST(aud_usd AS DOUBLE) as aud_usd,
        CAST(aud_inr as DOUBLE) as aud_inr
    FROM wa_fiscal.bronze.fx_rates
    where aud_usd is not null
)
SELECT 
    p.month_start,
    p.price_usd_per_tonne,
    f.aud_usd,
    f.aud_inr,
    p.price_usd_per_tonne / f.aud_usd as price_aud_per_tonne
from p join f on p.month_start = f.month_start
WHERE p.month_start >= '2015-01-01'
```

```python
%sql
select min(month_start), max(month_start), count(*) from wa_fiscal.silver.iron_ore_monthly 
```
